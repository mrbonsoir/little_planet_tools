# What do we do here?

We do the following:
+ load an equirectangular image *equiRecImage*
+ show the  *equiRecImage* image
+ create a *.pto* file
+ run the *nona* command using the *.pto* and *equiRecImage*
+ show the resulting little planet

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import sys
import colour as cs
import imageio as io
import importlib
import os

path_modules = "../my_modules/"
sys.path.append(path_modules)

import tools_for_little_planets as tflp
```

# Load an equirectangular image

```python
path_imgEqRec = "../data/pano_equirectangular.jpg"
imgEqRect = io.v2.imread(path_imgEqRec)

plt.imshow(imgEqRect)

print(imgEqRect.shape)
```

# Create config .pto file

```python
importlib.reload(tflp)

tflp.create_pto_config_file(path_imgEqRec, output_file_name = "template33",
output_width = 3000, output_height = 1500,
r_angle = 0, p_angle = 90, y_angle = 0,
output_distance = 300)
```

# Apply config /pto file

```python
path_nona = "/Applications/Hugin/tools_mac/nona"
cmd_ = "{0:s} -o outImgStereoGraphic -m JPEG template33.pto {1:s}".format(path_nona, path_imgEqRec)
print(cmd_)
os.system(cmd_)
```

# Display little planet image

```python
path_imgStereoGraphic = "outImgStereoGraphic.jpg"
imgStereoGraphic = io.v2.imread(path_imgStereoGraphic)

plt.imshow(imgStereoGraphic)

print(imgStereoGraphic.shape)
```
